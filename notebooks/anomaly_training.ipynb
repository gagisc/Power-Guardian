# pdu-power-guardian — Anomaly Model Training


```python
import sys; sys.path.insert(0,'../src')
import numpy as np, pandas as pd
from models.isolation_forest import PDUIsolationForest
from models.sequence_model import SequenceAnomalyDetector
```

```python
df = pd.read_csv('../data/samples/pdu_power.csv')
FEATS = ['eaton_power_w','eaton_current_a','apc_output_w','schneider_kw','schneider_load_pct']
X = df[FEATS].values
labels = df['label'].values
print(df.describe())
```

```python
clf = PDUIsolationForest(contamination=0.05).fit(X)
scores = clf.score(X)
import matplotlib.pyplot as plt
plt.figure(figsize=(12,3))
plt.plot(scores,label='IF score')
plt.scatter(np.where(labels==1),scores[labels==1],c='red',label='anomaly',zorder=5)
plt.legend(); plt.title('IsolationForest Scores'); plt.show()
```
